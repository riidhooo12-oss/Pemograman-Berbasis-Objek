Praktikum 01 Mendefinisikan Kelas Abstrak Sederhana

```python
from abc import ABC, abstractmethod

class KendaraanAbstrak(ABC):

    def __init__(self, merk):
        self.merk = merk
        print(f"Inisialisasi Kendaraan: {self.merk}")

    def info_merk(self):
        print(f"Merk kendaraan: {self.merk}")

    @abstractmethod
    def start_mesin(self):
        pass

    @abstractmethod
    def stop_mesin(self):
        pass


class Mobil(KendaraanAbstrak):

    def start_mesin(self):
        print(f"Mesin {self.merk} dinyalakan")

    def stop_mesin(self):
        print(f"Mesin {self.merk} dimatikan")


mobil = Mobil("Toyota")

mobil.start_mesin()
mobil.info_merk()
mobil.stop_mesin()
```

Praktikum 03 Membuat Subclass Konkret Alat Pembayaran

```python
from abc import ABC, abstractmethod

class AlatPembayaranAbstrak(ABC):

    def __init__(self,nama):
        self.nama=nama

    def info(self):
        print("Metode:",self.nama)

    @abstractmethod
    def proses_pembayaran(self,jumlah):
        pass


class KartuKredit(AlatPembayaranAbstrak):

    def __init__(self,nomor,nama):
        super().__init__("Kartu Kredit")
        self.nomor=nomor
        self.nama_pemilik=nama

    def proses_pembayaran(self,jumlah):
        print(f"Pembayaran {jumlah} menggunakan kartu kredit berhasil")
        return True


class DompetDigital(AlatPembayaranAbstrak):

    def __init__(self,nohp,saldo):

        super().__init__("Dompet Digital")

        self.nohp=nohp
        self.saldo=saldo

    def proses_pembayaran(self,jumlah):

        if jumlah<=self.saldo:

            self.saldo-=jumlah

            print("Pembayaran berhasil")
            print("Sisa saldo:",self.saldo)

            return True

        else:

            print("Saldo tidak cukup")
            return False


kk=KartuKredit("123456","Budi")

gopay=DompetDigital("08123456789",100000)

kk.proses_pembayaran(50000)

gopay.proses_pembayaran(75000)
```

Praktikum 04 Polimorfisme Dokumen

```python
from abc import ABC,abstractmethod

class DokumenAbstrak(ABC):

    def __init__(self,nama):
        self.nama=nama

    @abstractmethod
    def cetak(self):
        pass

    @abstractmethod
    def simpan(self):
        pass


class DokumenTeks(DokumenAbstrak):

    def cetak(self):
        print("Mencetak dokumen teks")

    def simpan(self):
        print("Menyimpan file txt")


class Spreadsheet(DokumenAbstrak):

    def cetak(self):
        print("Mencetak spreadsheet")

    def simpan(self):
        print("Menyimpan file xlsx")


def proses_dokumen(data):

    for x in data:

        x.cetak()

        x.simpan()


d1=DokumenTeks("laporan.txt")

d2=Spreadsheet("nilai.xlsx")

data=[d1,d2]

proses_dokumen(data)
```

Praktikum 05 Metode Konkret dan Abstrak

```python
from abc import ABC,abstractmethod
import math

class BangunDatarAbstrak(ABC):

    def __init__(self,nama):

        self._nama=nama


    @property
    def nama(self):
        return self._nama


    def info_lengkap(self):

        print("Nama:",self.nama)

        print("Luas:",self.hitung_luas())

        print("Keliling:",self.hitung_keliling())


    @abstractmethod
    def hitung_luas(self):
        pass


    @abstractmethod
    def hitung_keliling(self):
        pass



class Lingkaran(BangunDatarAbstrak):

    def __init__(self,r):

        super().__init__("Lingkaran")

        self.r=r


    def hitung_luas(self):

        return math.pi*self.r**2


    def hitung_keliling(self):

        return 2*math.pi*self.r


ling=Lingkaran(10)

ling.info_lengkap()
```

Praktikum 06 Properti Abstrak Elemen Grafis

```python
from abc import ABC,abstractmethod

class ElemenGrafis(ABC):

    def __init__(self,id,warna):

        self._id=id

        self.warna=warna


    @property
    def id_elemen(self):
        return self._id


    @property
    @abstractmethod
    def posisi(self):
        pass


    @posisi.setter
    @abstractmethod
    def posisi(self,data):
        pass


    @abstractmethod
    def gambar(self):
        pass


class Kotak(ElemenGrafis):

    def __init__(self,id,warna,x=0,y=0):

        super().__init__(id,warna)

        self.x=x

        self.y=y


    @property
    def posisi(self):

        return (self.x,self.y)


    @posisi.setter
    def posisi(self,data):

        self.x=data[0]

        self.y=data[1]


    def gambar(self):

        print(f"Kotak {self.id_elemen}")
        print("Posisi:",self.posisi)
        print("Warna:",self.warna)



k=Kotak("K1","Merah")

print(k.posisi)

k.posisi=(50,70)

print(k.posisi)

k.gambar()
```

Penugasan Program Abstract Base Class Senjata

```python
from abc import ABC, abstractmethod

# Kelas Abstrak
class Senjata(ABC):
    def __init__(self, nama):
        self.nama = nama

    def info_nama(self):
        print("Nama Senjata:", self.nama)

    @abstractmethod
    def serang(self):
        pass

    @property
    @abstractmethod
    def ammo(self):
        pass


# Kelas Pedang
class Pedang(Senjata):
    def __init__(self, nama, panjang_bilah):
        super().__init__(nama)
        self.panjang_bilah = panjang_bilah
        self._daya_tahan = 100

    def serang(self):
        print(f"Pedang {self.nama} menebas dengan bilah sepanjang {self.panjang_bilah} cm!")

    @property
    def ammo(self):
        return self._daya_tahan


# Kelas Panah
class Panah(Senjata):
    def __init__(self, nama, jumlah_anak_panah):
        super().__init__(nama)
        self._jumlah_anak_panah = jumlah_anak_panah

    def serang(self):
        if self._jumlah_anak_panah > 0:
            self._jumlah_anak_panah -= 1
            print(f"Panah {self.nama} melesat! Sisa anak panah: {self._jumlah_anak_panah}")
        else:
            print(f"Amunisi Panah {self.nama} habis!")

    @property
    def ammo(self):
        return self._jumlah_anak_panah

    @ammo.setter
    def ammo(self, isi):
        self._jumlah_anak_panah += isi


# Program utama
if __name__ == "__main__":

    pedang1 = Pedang("Excalibur",120)
    panah1 = Panah("Crossbow",3)

    print("=== Pedang ===")
    pedang1.info_nama()
    pedang1.serang()
    print("Daya tahan:",pedang1.ammo)

    print("\n=== Panah ===")
    panah1.info_nama()

    panah1.serang()
    panah1.serang()
    panah1.serang()
    panah1.serang()

    print("Sisa ammo:",panah1.ammo)

    panah1.ammo=5

    print("Ammo setelah isi ulang:",panah1.ammo)
```
